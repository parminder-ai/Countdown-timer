```python

# ==========================================
# DATA SCIENCE PROJECT
# MOVIE RATING ANALYSIS USING PYTHON
# ==========================================

# STEP 1: Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set graph style
sns.set_theme(style="whitegrid")

# STEP 2: Create Movie Dataset

data = {
    'Movie': [
        'Avatar', 'Titanic', 'Inception', 'Interstellar',
        'The Dark Knight', 'Avengers', 'Joker',
        'Frozen', 'Toy Story', 'The Lion King',
        'Spider-Man', 'Iron Man', 'The Matrix',
        'Gladiator', 'Coco', 'Up', 'Finding Nemo',
        'Dangal', '3 Idiots', 'Bahubali'
    ],

    'Genre': [
        'Action', 'Romance', 'Sci-Fi', 'Sci-Fi',
        'Action', 'Action', 'Drama',
        'Animation', 'Animation', 'Animation',
        'Action', 'Action', 'Sci-Fi',
        'Action', 'Animation', 'Animation',
        'Animation', 'Drama', 'Comedy', 'Action'
    ],

    'Year': [
        2009, 1997, 2010, 2014, 2008,
        2012, 2019, 2013, 1995, 1994,
        2017, 2008, 1999, 2000, 2017,
        2009, 2003, 2016, 2009, 2015
    ],

    'Rating': [
        7.9, 7.9, 8.8, 8.6, 9.0,
        8.0, 8.3, 7.4, 8.3, 8.5,
        7.4, 7.9, 8.7, 8.5, 8.4,
        8.3, 8.2, 8.3, 8.4, 8.0
    ],

    'Votes': [
        150000, 180000, 200000, 190000,
        250000, 170000, 160000, 130000,
        140000, 175000, 145000, 165000,
        185000, 155000, 120000, 110000,
        125000, 135000, 195000, 115000
    ]
}

df = pd.DataFrame(data)

# STEP 3: Display Dataset

print("MOVIE RATING DATASET")
print(df)

print("\nDataset Information:")
print(df.info())

# STEP 4: Basic Data Analysis

print("\nTotal Number of Movies:", len(df))

print("Average Movie Rating:",
      round(df['Rating'].mean(), 2))

print("Highest Movie Rating:",
      df['Rating'].max())

print("Lowest Movie Rating:",
      df['Rating'].min())

# STEP 5: Find Top 10 Movies

top_movies = df.sort_values(
    by='Rating', ascending=False
).head(10)

print("\nTOP 10 HIGHEST RATED MOVIES")
print(top_movies[['Movie', 'Rating']])

# STEP 6: Genre Wise Average Rating

genre_rating = df.groupby('Genre')['Rating'].mean()

print("\nAVERAGE RATING BY GENRE")
print(genre_rating.round(2))

# STEP 7: Count Movies by Genre

genre_count = df['Genre'].value_counts()

print("\nNUMBER OF MOVIES BY GENRE")
print(genre_count)

# STEP 8: Graph 1 - Top 10 Movies

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_movies,
    x='Rating',
    y='Movie'
)

plt.title('Top 10 Highest Rated Movies')
plt.xlabel('Movie Rating')
plt.ylabel('Movie Name')
plt.tight_layout()
plt.show()

# STEP 9: Graph 2 - Average Rating by Genre

plt.figure(figsize=(9, 5))

genre_rating.sort_values().plot(
    kind='bar',
    edgecolor='black'
)

plt.title('Average Movie Rating by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# STEP 10: Graph 3 - Number of Movies by Genre

plt.figure(figsize=(8, 5))

genre_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Movies by Genre')
plt.ylabel('')
plt.tight_layout()
plt.show()

# STEP 11: Graph 4 - Ratings Distribution

plt.figure(figsize=(8, 5))

sns.histplot(
    df['Rating'],
    bins=8,
    kde=True
)

plt.title('Distribution of Movie Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Movies')
plt.tight_layout()
plt.show()

# STEP 12: Rating and Votes Relationship

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Votes',
    y='Rating',
    hue='Genre',
    s=100
)

plt.title('Relationship Between Votes and Ratings')
plt.xlabel('Number of Votes')
plt.ylabel('Movie Rating')
plt.tight_layout()
plt.show()

# STEP 13: Final Project Summary

print("\n========== PROJECT SUMMARY ==========")

print("Total Movies Analysed:", len(df))

print("Overall Average Rating:",
      round(df['Rating'].mean(), 2))

best_movie = df.loc[df['Rating'].idxmax()]

print("Highest Rated Movie:",
      best_movie['Movie'])

print("Highest Rating:",
      best_movie['Rating'])

best_genre = genre_rating.idxmax()

print("Highest Average Rated Genre:",
      best_genre)

print("\nPROJECT COMPLETED SUCCESSFULLY!")

# STEP 14: Save Dataset as CSV

df.to_csv('movie_rating_analysis.csv', index=False)

print("\nDataset saved as movie_rating_analysis.csv")
```
